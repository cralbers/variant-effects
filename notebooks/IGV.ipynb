```python
import igv_notebook
igv_notebook.init()
```

```python
LMNA_START = 156_082_572
LMNA_END = 156_140_081
gene_symbol = "LMNA"

locus = 'chr1:155014711-157240081'
```

```python
path = '/Users/coraalbers/Documents/BSGP/Lancaster_rotation/variant-effects/osc/data_sync/'
track_list = []
beds={'pre':'predicted_ccre_like_regions_pre_LV_128res.bed',
      'central':'predicted_ccre_like_regions_central_LV_128res.bed',
      'post':'predicted_ccre_like_regions_post_LV_128res.bed',
      'pre-collapsed':'predicted_reg_regions_pre_indivtracks_LV_128res_collapsed.bed',
      'central-collapsed':'predicted_reg_regions_central_indivtracks_LV_128res_collapsed.bed',
      'post-collapsed':'predicted_reg_regions_post_indivtracks_LV_128res_collapsed.bed'}
colors=['orange','green','gray', 'red', 'blue', 'purple']
i=0
#add a bam for each of the files above
for b in beds:
    d = {"name": b,
        "url":path+beds[b],
        "type": "annotation",
         "displayMode":"EXPANDED",
         "height":50,
         "removable":True,
         "color":colors[i],
        "indexed": True }
    track_list.append(d)
    i+=1

igv_browser= igv_notebook.Browser(
    {
        "genome": "hg38",
        "locus": locus,
        "tracks": track_list,
        "roi": {
                "name": "LMNA",
                "color": "red",
                "features": [
                    {
                        "chr": "chr1",
                        "start": LMNA_START,
                        "end": LMNA_END
                    }   ,
                ]
            }
        }
    
)


```
